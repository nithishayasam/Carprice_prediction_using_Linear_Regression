ONE HOT ENCODING

```python
encoded = pd.get_dummies(X ,columns=['model', 'transmission', 'fuelType'],drop_first=True)
```

```python
encoded = encoded.astype(int)
```

```python
encoded
```

label encoding

```python
from sklearn.preprocessing import LabelEncoder
```

```python
columns_to_encode = ['model', 'transmission', 'fuelType']
```

```python
xlabel = X.copy()
```

```python
label_encoders = {}
for col in columns_to_encode:
    le = LabelEncoder()
    xlabel[col] = le.fit_transform(xlabel[col])
    label_encoders[col] = le
```

using standard scaler to scale the numerical features in the dataset for the one hot encoded data

```python
from sklearn.preprocessing import StandardScaler

columns = ['year', 'mileage',"tax","mpg", 'engineSize']
scaler = StandardScaler()
```

```python
encoded[columns] = scaler.fit_transform(encoded[columns])
```

```python
encoded
```

using StandardScaler to scale the numerical features in the dataset for the label encoded data

```python
xlabel.columns
```

```python
xlabel[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']] = scaler.fit_transform(xlabel[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg','engineSize']])
```

```python
xlabel
```

checking model accuracy using one hot encoding data

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score,mean_absolute_error
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(encoded, Y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, Y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
Y_test
```

```python
r2=  r2_score(Y_test, y_pred)
print(" One Hot Encoded Data R-squared:", r2)
```

```python
n = X_test.shape[0]
p = X_test.shape[1]
```

```python
adjusted_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)
print(" One Hot Encoded Data R-squared:", r2)
print(" One Hot Encoded Data Adjusted R-squared:", adjusted_r2)
```

checking model accuracy using label encoded data

```python
x_train, x_test, y_train, y_test = train_test_split(xlabel, Y, test_size=0.2, random_state=42)
```

```python
model2 = LinearRegression()
```

```python
model2.fit(x_train, y_train)
```

```python
y_pred2 = model2.predict(x_test)
```

```python
y_test
```

```python
y_train
```

```python
R2 = r2_score(y_test, y_pred2)
print(" Label Encoded Data R-squared:", R2)
```

```python
n1 = x_test.shape[0]
p1 = x_test.shape[1]
```

```python
adjusted_R2 = 1 - (1 - R2) * (n1 - 1) / (n1 - p1 - 1)
print(" Label Encoded Data Adjusted R-squared:", adjusted_R2)
```

```python
print("One Hot Encoding R-squared:", r2)
print("One Hot Encoding Adjusted R-squared:", adjusted_r2)
print("Label Encoding R-squared:", R2)
print("Label Encoding Adjusted R-squared:", adjusted_R2)
```

```python
if r2 > R2:
    print("One Hot Encoding performs better for this dataset.")
else:
    print("Label Encoding performs better for this dataset.")
```
